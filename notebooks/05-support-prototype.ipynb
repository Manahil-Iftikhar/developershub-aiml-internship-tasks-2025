## Pretrained inference baseline

This reproduces the kind of inference present in the submission. It does **not** fine-tune the model.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

model_id = 'tiiuae/falcon-rw-1b'
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)
generator = pipeline('text-generation', model=model, tokenizer=tokenizer)

def baseline_response(question):
    if not question.strip():
        raise ValueError('Provide a question.')
    prompt = f'Provide a brief, empathetic reply without diagnosis or treatment advice.\nUser: {question}\nReply:'
    return generator(prompt, max_new_tokens=80, do_sample=False,
                     return_full_text=False, pad_token_id=tokenizer.eos_token_id)[0]['generated_text'].strip()
```

```python
baseline_response('I feel nervous about learning a new skill.')
```
