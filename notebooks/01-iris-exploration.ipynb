## Inspect the bundled Iris dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
frame = iris.frame.rename(columns={'target': 'species_id'})
frame['species'] = frame['species_id'].map(dict(enumerate(iris.target_names)))
features = list(iris.feature_names)
print('Rows:', len(frame))
print('Missing values:', frame[features].isna().sum().sum())
print(frame['species'].value_counts())
frame[features].describe()
```

## Compare petal measurements

```python
sns.set_theme(style='whitegrid', palette='colorblind')
fig, ax = plt.subplots(figsize=(9, 5))
sns.scatterplot(data=frame, x=features[2], y=features[3], hue='species', ax=ax)
ax.set_title('Iris species by petal measurements')
fig.tight_layout()
fig.savefig(ARTIFACTS / 'iris-petals.png', dpi=160)
plt.show()
```

## Inspect distributions and spread

```python
frame[features].hist(figsize=(10, 7), bins=15, edgecolor='white')
plt.tight_layout()
plt.show()
fig, axes = plt.subplots(1, 4, figsize=(14, 4))
for column, ax in zip(features, axes):
    sns.boxplot(data=frame, x='species', y=column, ax=ax)
    ax.tick_params(axis='x', rotation=35)
fig.tight_layout()
plt.show()
```

## Findings

Setosa separates clearly on petal measurements in this sample. Versicolor and virginica overlap more. These are descriptive findings from a small, balanced reference dataset; they do not establish generalization to new collection conditions.
