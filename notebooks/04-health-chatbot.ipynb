## Load an experimental text model

This cell downloads a checkpoint on its first run. Generated responses must be reviewed.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

model_id = 'google/flan-t5-small'
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)
model.eval()

def ask_general_question(question):
    if not question.strip():
        raise ValueError('Provide a non-empty question.')
    prompt = ('Give general educational health information. Do not diagnose or prescribe. '
              'State uncertainty and suggest professional help when appropriate. '
              f'Question: {question}\nAnswer:')
    inputs = tokenizer(prompt, return_tensors='pt', truncation=True, max_length=512)
    with torch.inference_mode():
        output = model.generate(**inputs, max_new_tokens=120, do_sample=False)
    return tokenizer.decode(output[0], skip_special_tokens=True)
```

## Inspect one response

```python
ask_general_question('What habits can affect sleep quality?')
```
