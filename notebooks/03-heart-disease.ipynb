## Load the reviewed source

Convert original labels and categorical codes without fitting preprocessing on holdout rows.


```python
from portfolio.heart import load_cleveland, run
path = DATA / 'processed.cleveland.data'
frame, missing = load_cleveland(path)
print(frame.shape, missing)
frame.head()
```

## Fit, select on validation data, and evaluate on the test set

```python
import json
report = run(path, ARTIFACTS / 'heart-cleveland')
print(json.dumps({'selected_model': report['selected_model'], 'test': report['test']}, indent=2))
```

## Inspect errors

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import ConfusionMatrixDisplay
matrix = report['test']['selected_model']['confusion_matrix']
fig, ax = plt.subplots(figsize=(5, 4))
ConfusionMatrixDisplay(np.asarray(matrix), display_labels=['Class 0', 'Class 1']).plot(ax=ax, colorbar=False)
ax.set_title('Cleveland held-out test errors')
fig.tight_layout()
plt.show()
```
